```python
import pandas as pd
import datetime

# load both files with full path
df_flipkart = pd.read_csv("/Users/sahasra/Desktop/ecommerce_project/flipkart_raw_21Apr2026.csv")
df_shopsy   = pd.read_csv("/Users/sahasra/Desktop/ecommerce_project/shopsy_raw_21Apr2026.csv")

print(f"Flipkart rows : {len(df_flipkart)}")
print(f"Shopsy rows   : {len(df_shopsy)}")

# combine
df_combined = pd.concat([df_flipkart, df_shopsy], ignore_index=True)

print(f"Combined rows : {len(df_combined)}")
print(f"\nPlatforms:\n{df_combined['platform'].value_counts()}")
print(f"\nCategories:\n{df_combined['category'].value_counts()}")

# save master file in same folder
master_file = "/Users/sahasra/Desktop/ecommerce_project/master_raw.csv"
df_combined.to_csv(master_file, index=False, encoding="utf-8-sig")

print(f"\nMaster dataset saved!")
print(f"Total rows: {len(df_combined)}")
```
